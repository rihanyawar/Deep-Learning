```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("Datasets/DailyDelhiClimate.csv")
```

```python
df.head()
```

```python
df= df[['date','meantemp']]
```

```python
df.head()
```

```python
df.shape
```

```python
plt.scatter(df['date'], df['meantemp'],color='red')
```

```python
x1= df['meantemp'].ewm(alpha=0.9).mean()
x2= df['meantemp'].ewm(alpha=0.5).mean()
x3= df['meantemp'].ewm(alpha=0.1).mean()
```

```python
x1
```

```python
x2
```

```python
x3
```

```python
df['ewma1']= x1
df['ewma2']= x2
df['ewma3']= x3
df.head()
```

```python
plt.scatter(df['date'], df['meantemp'],color='red')
plt.plot(df['date'],df['ewma1'], color='black', label= 'alpha=0.9')
plt.legend()
plt.show()
```

```python
plt.scatter(df['date'], df['meantemp'],color='red')
plt.plot(df['date'],df['ewma2'], color='black', label= 'alpha=0.5')
plt.legend()
plt.show()
```

```python
plt.scatter(df['date'], df['meantemp'],color='red')
plt.plot(df['date'],df['ewma3'], color='black', label= 'alpha=0.1')
plt.legend()
plt.show()
```
